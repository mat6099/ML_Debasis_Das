# Visualizing Data

## matplotlib

**matplotlib**, the most common Python library for charts.

- Good for: simple **bar charts, line charts, scatterplots**.
- Not the best for: fancy interactive charts on websites.

We use its `pyplot` module, and by tradition we call it `plt`:


`from matplotlib import pyplot as plt`  or  `import matplotlib.pyplot as plt`

**How pyplot works:** you build the chart **step by step** (add line → add title → add labels), and at the end you either:
- **show** it → `plt.show()`
- **save** it → `plt.savefig("file.png")`

If matplotlib is not installed, run: `pip install matplotlib`

```python
# import pyplot from matplotlib
import matplotlib.pyplot as plt
```

## Line chart

A **line chart** is best for showing **how something changes over time** (a trend).

**Command:** `plt.plot(x_values, y_values)`

Useful options:
| Option | Meaning | Example |
|---|---|---|
| `color` | line colour | `color="green"` |
| `marker` | dot on each point | `marker="o"` |
| `linestyle` | solid / dashed / dotted | `linestyle="solid"` |

**Example:** India's population at every Census (in crore).

```python
census_years = [1951, 1961, 1971, 1981, 1991, 2001, 2011]
population_crore = [36.1, 43.9, 54.8, 68.3, 84.6, 102.9, 121.1]

# Step 1: draw the line
plt.plot(census_years, population_crore, color="blue", marker="o", linestyle="solid", linewidth=2)

# Step 2: add title and labels
plt.title("India's Population by Census Year")
plt.xlabel("Census year")
plt.ylabel("Population (crore)")

# Step 3: show the chart
plt.show()
```

## Bar chart

A **bar chart** is best when you want to **compare a quantity across different categories** (items, groups, names).

**Command:** `plt.bar(x_positions, heights)`

**Example:** Cups of tea sold at a college canteen tea stall during one week.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
tea_cups = [120, 135, 110, 150, 180, 90]

plt.bar(days, tea_cups, color="blue", width=0.6)

plt.title("Tea Cups Sold at the Canteen This Week")
plt.xlabel("Day")
plt.ylabel("Number of cups")
plt.show()
```

```python
subjects = ["Mathematics", "Computer Science", "Physics", "Economics", "English Literature"]
students = [42, 65, 30, 25, 18]

# sort from smallest to biggest so the biggest bar is on top
pairs = sorted(zip(students, subjects))
counts, names = zip(*pairs)            # unzip (we learnt this in Chapter 2!)

plt.barh(names, counts, color="blue", height=0.6)
plt.title("Students Choosing Each Elective")
plt.xlabel("Number of students")
plt.grid(axis="y", visible=False)      # keep only vertical grid lines
plt.show()
```

## Histogram

A **histogram** shows **how values are spread out** (their *distribution*).

**Example:** Maths marks of 20 students, grouped into buckets of 10 (0–9, 10–19, … 90–99, 100).

Steps:
1. Find the bucket of each mark: `mark // 10 * 10` (ex. 87 → 80).
2. Count marks in each bucket using `Counter` (from Chapter 2).
3. Draw a bar chart of the counts.

```python
from collections import Counter

marks = [78, 92, 85, 67, 45, 88, 73, 95, 81, 59,
         100, 76, 84, 62, 91, 70, 33, 87, 79, 68]

def bucket(mark):
    """Returns the starting value of the 10-mark group. Example: 87 -> 80"""
    return mark // 10 * 10

histogram = Counter(bucket(m) for m in marks)
print(sorted(histogram.items())) 
```

```python
plt.bar(list(histogram.keys()),      # x position = bucket (in Python 3, wrap in list())
        list(histogram.values()),    # height = number of students
        width=8,                     # width 8 leaves a small gap (buckets are 10 apart)
        color="blue")

plt.xticks([10 * i for i in range(11)])   # labels at 0, 10, 20, ..., 100
plt.axis([-5, 105, 0, 8])                 # x from -5 to 105, y from 0 to 8
plt.title("Distribution of Maths Marks")
plt.xlabel("Marks group (80 means 80–89)")
plt.ylabel("Number of students")
plt.show()
```

Most students scored in the **70s and 80s**. Only a few scored below 50 – these students may need extra help.

**Notes:**
- `plt.axis([xmin, xmax, ymin, ymax])` sets the range of both axes.
- `plt.xticks([...])` decides where the labels on the x-axis go.
- The book shifts each bar left by 4 (`x - 4`) because old matplotlib placed bars from the left edge. Today bars are centred, so no shift is needed.

#### Shortcut: `plt.hist()`
matplotlib can also do the bucketing and counting for you:

```python
plt.hist(marks, bins=range(0, 111, 10), color="blue", edgecolor="white", linewidth=2)
plt.title("Same Marks Using plt.hist()")
plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.show()
```

In a bar chart, the **y-axis must start at 0**.

If it starts somewhere else, a very small difference can **look huge**. This is a common trick in advertisements and news graphics.

**Example:** A shop's visitors went from 500 in 2025 to 505 in 2026 – just **1% more**. Let's draw it both ways, side by side.

(We use `plt.subplots(1, 2)` to put two charts next to each other.)

```python
years = ["2025", "2026"]
visitors = [500, 505]

fig, (left, right) = plt.subplots(1, 2, figsize=(10, 4))

# ❌ Misleading: y-axis starts at 499
left.bar(years, visitors, color="orange", width=0.5)
left.set_ylim(499, 506)
left.set_title("Misleading: 'Huge' Growth!")
left.set_ylabel("Shop visitors")

# ✅ Honest: y-axis starts at 0
right.bar(years, visitors, color="blue", width=0.5)
right.set_ylim(0, 550)
right.set_title("Honest: Almost No Change")
right.set_ylabel("Shop visitors")

plt.tight_layout()  # adjusts spacing between subplots to prevent overlap   
plt.show()
```

**Always check where the y-axis starts** – both when making charts and when reading charts made by others.

(For **line charts**, the y-axis doesn't always need to start at 0, because we are looking at the *trend*, not the bar length. But be careful and honest.)

## Line chart

You can call `plt.plot()` **many times** to draw several lines on the same chart.

If you give each line a `label`, then `plt.legend()` makes a **legend** (the little box that says which line is which) automatically.

**Short style codes** for colour + line type:
| Code | Meaning |
|---|---|
| `'g-'` | green, solid line |
| `'r-.'` | red, dash-dot line |
| `'b:'` | blue, dotted line |
| `'k--'` | black, dashed line |

**Example:** Mobile data (GB) used each month by three family members.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
dad    = [ 8,  9,  8, 10,  9, 11]
mom    = [12, 14, 13, 15, 18, 17]
ricky   = [25, 28, 35, 42, 55, 48]   # summer vacation in May!

plt.plot(months, ricky, color="blue",   marker="o", linestyle="-",  linewidth=2, label="Ricky")
plt.plot(months, mom,  color="orange", marker="s", linestyle="--", linewidth=2, label="Mom")
plt.plot(months, dad,  color="aqua",   marker="^", linestyle=":",  linewidth=2, label="Dad")

plt.legend(loc="upper left")         # we gave labels, so legend is free
plt.title("Monthly Mobile Data Usage")
plt.xlabel("Month")
plt.ylabel("Data used (GB)")
plt.ylim(0, 60)                      # set y-axis from 0 to 60
plt.show()
```

## Scatterplot

A **scatterplot** shows the **relationship between two sets of numbers** that come in pairs.

Each dot = one person/item, with:
- x-position = first value
- y-position = second value

**Command:** `plt.scatter(x_values, y_values)`

To put a name next to each dot, use `plt.annotate(text, xy=(x, y))`.

**Example:** For 9 students – hours studied in the week before the exam vs marks scored.

```python
names         = ["Aman", "Bina", "Chetan", "Diya", "Esha", "Farhan", "Gita", "Harsh", "Isha"]
hours_studied = [2, 5, 8, 3, 10, 6, 1, 7, 9]
marks_scored  = [45, 62, 80, 50, 92, 70, 38, 74, 85]

plt.scatter(hours_studied, marks_scored, color="blue", s=60, edgecolor="white", linewidth=1.5)

# write each student's name next to their dot
for name, h, m in zip(names, hours_studied, marks_scored):
    plt.annotate(name,
                 xy=(h, m),              # the point
                 xytext=(6, -4),         # move the text a little right
                 textcoords="offset points")

plt.title("Hours Studied vs Marks Scored")
plt.xlabel("Hours studied in the week before the exam")
plt.ylabel("Marks scored (out of 100)")
plt.show()
```
