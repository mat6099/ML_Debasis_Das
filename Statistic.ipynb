# Statistics

## Central Tendency

**Central tendency** is a single number that tells us where the centre of the data is. In other words, it gives the typical value of the data.

For example, instead of reading the marks of all 60 students, the teacher can say "the class average is 72". That one number gives a quick idea of the whole class.

### measures of central tendency

| Measure | Meaning | How to find it |
|---|---|---|
| **Mean** (Average) | Sum of all values ÷ number of values | Add everything, then divide |
| **Median** | The middle value | Sort the data, then pick the middle one |
| **Mode** | The value that appears most often | Count which value repeats the most |

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Example list of height, weight, and age of 5 individuals

height = [160, 150, 170, 140, 120]
weight = [60, 50, 70, 40, 30]
age = [25, 30, 22, 28, 23]
```

```python
# Mean of height, weight, and age

def mean(data):
    return sum(data) / len(data)

mean_height = mean(height)
mean_weight = mean(weight)
mean_age = mean(age)

print("Mean Height:", mean_height)
print("Mean Weight:", mean_weight)
print("Mean Age:", mean_age)

# Another way to calculate mean using numpy
mean_height_np = np.mean(height)
mean_weight_np = np.mean(weight)
mean_age_np = np.mean(age)
print("Mean Height (NumPy):", mean_height_np)
print("Mean Weight (NumPy):", mean_weight_np)
print("Mean Age (NumPy):", mean_age_np)
```
