## Functions

A **function** takes zero or more inputs and gives back an output. We define it with `def`.

Key points:
- A **docstring** (text in triple quotes) explains what the function does.
- Functions are **first-class** – you can store them in variables and pass them to other functions.
- **Lambda** = a short, one-line function without a name.
- Parameters can have **default values**.
- You can pass arguments **by name**.

```python
def add_gst(price):
    """Returns the price after adding 18% GST."""
    return price * 1.18

print(add_gst(1000))
```

```python
# Functions are first-class: we can pass one function into another
def apply_to_100(f):
    """Calls the function f with 100 as input."""
    return f(100)

my_func = add_gst         
print(apply_to_100(my_func))   

# Lambda – a small nameless function
print(apply_to_100(lambda x: x - 10))

# Note: assigning a lambda to a name works, but 'def' is preferred
def half(x): return x / 2
```

```python
# Default arguments
def greet(name="Guest"):
    print(f"Namaste, {name}!")

greet("Debasis Das")
greet()          

# Arguments by name
def emi_left(total=0, paid=0):
    return total - paid

print(emi_left(50000, 20000))
print(emi_left(paid=5000))  
```

## Exceptions

When something goes wrong, Python **raises an exception**. If we don't handle it, the program crashes.

We handle it using `try` and `except`. In Python, using exceptions to keep code clean is perfectly normal.

```python
def per_head_cost(total_bill, people):
    try:
        return total_bill / people
    except ZeroDivisionError:
        print("Cannot divide the bill among 0 people!")
        return None

print(per_head_cost(1200, 4))
print(per_head_cost(1200, 0))
```

## Lists

A **list** is an ordered collection – the most important data structure in Python. (Like an array in other languages, but more powerful.)

| Operation | Example |

## Tuples

A **tuple** is like a list, but it **cannot be changed** (it is *immutable*). We write it with round brackets `( )`.

Tuples are very handy for:
- Returning **multiple values** from a function.
- **Swapping** variables in one line.

```python
train = ("12951", "Bhubaneswar Rajdhani")   # tuple
try:
    train[1] = "Duronto"
except TypeError:
    print("Cannot modify a tuple!")

def min_and_max(temps):
    return min(temps), max(temps)      # returns a tuple

low, high = min_and_max([28, 31, 35, 33, 29])
print(f"Lowest: {low}°C, Highest: {high}°C")

# Swapping – the Pythonic way
a, b = "tea", "coffee"
a, b = b, a
print(a, b)   # coffee tea
```

## Dictionaries

A **dictionary** stores **key → value** pairs. You can find a value very fast using its key.

- Look up: `d[key]` (gives `KeyError` if key is missing)
- Safe look up: `d.get(key, default)`
- Check key: `key in d`
- Add / update: `d[key] = value`
- `.keys()`, `.values()`, `.items()` give all keys, values, or pairs.
- Keys must be **immutable** (string, number, tuple – **not** list).

```python
pin_codes = {"Kolkata": 700001, "Mumbai": 400001, "Chennai": 600001}

print(pin_codes["Mumbai"])

try:
    print(pin_codes["Jaipur"])
except KeyError:
    print("No PIN code saved for Jaipur!")

print("Chennai" in pin_codes)              # True
print(pin_codes.get("Jaipur", "Unknown"))  # Unknown – no error

pin_codes["Jaipur"] = 302001               # add new
pin_codes["Kolkata"] = 700091              # update existing
print(len(pin_codes), pin_codes)
```

```python
# Dictionaries are great for structured data – like one record
movie = {
    "title": "Dhurandar",
    "director": "Aditya Dhar",
    "year": 2023,
    "languages": ["Hindi", "English", "Tamil", "Telugu"],
}

print(movie.keys())
print(movie.values())
for key, value in movie.items():
    print(f"{key:>10} : {value}")

print("director" in movie)  # True
```

## `defaultdict`

Suppose we want to **count words**. With a normal dict we must first check *"is this key already there?"* every time. That is boring.

A **`defaultdict`** solves this. If a key is missing, it automatically creates it with a default value:
- `defaultdict(int)` → default `0`
- `defaultdict(list)` → default `[]`
- `defaultdict(dict)` → default `{}`

```python
sentence = "chai pe charcha chai pe samosa chai"
words = sentence.split()

# Way 1: normal dict – need an if/else check
counts = {}
for w in words:
    if w in counts:
        counts[w] += 1
    else:
        counts[w] = 1
print("Normal dict:", counts)

# Way 2: dict.get
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1
print("Using get:  ", counts)

# Way 3: defaultdict – cleanest
from collections import defaultdict
counts = defaultdict(int)
for w in words:
    counts[w] += 1
print("defaultdict:", dict(counts))
```

```python
# defaultdict(list) – group students by their city
students = [("Amit", "Delhi"), ("Ricky", "Bhubaneswar"), ("Ravi", "Delhi"), ("Anu", "Pune"), ("Kiran", "Goa")]

by_city = defaultdict(list)
for name, city in students:
    by_city[city].append(name)
print(dict(by_city))

# defaultdict(dict) – nested data without checks
profile = defaultdict(dict)
profile["Amit"]["city"] = "Delhi"
print(dict(profile))
```

### `Counter`

**`Counter`** turns a list of items into a dictionary of **item → count**. It is the easiest way to make a frequency table (histogram).

`most_common(n)` gives the top `n` items.

```python
from collections import Counter

# Class poll: "Which is your favourite sweet?"
votes = ["Rasgulla", "Jalebi", "Ladoo", "Rasgulla", "Gulab Jamun",
         "Jalebi", "Rasgulla", "Ladoo", "Rasgulla"]
vote_count = Counter(votes)
print(vote_count)

for sweet, count in vote_count.most_common(2):
    print(sweet, count)

# Word counting in one line
print(Counter(sentence.split()).most_common(1))
```

## Sets

A **set** is a collection of **unique** items (no duplicates).

We use sets for two reasons:
1. **Fast membership check** – `x in my_set` is much faster than `x in my_list` for big data.
2. **Finding distinct items** – remove duplicates quickly.

```python
visited = set()
visited.add("Darjeeling")
visited.add("Goa")
visited.add("Goa")          # duplicate – ignored
print(visited, len(visited))

# 1. Fast membership check
stopwords = set(["a", "an", "the", "is", "of", "and"])
print("the" in stopwords)    # very fast

# 2. Distinct items
roll_calls = ["Riya", "Sam", "Riya", "Arjun", "Sam", "Riya"]
print("Total entries:", len(roll_calls))
print("Unique students:", len(set(roll_calls)), set(roll_calls))

# Bonus: set operations
mumbai_fans = {"Rohit", "Arjun", "Meera"}
chennai_fans = {"Meera", "Karthik"}
print("Fans of both teams:", mumbai_fans & chennai_fans)
```

## Control flow

- `if` / `elif` / `else` for decisions
- One-line if-else (ternary): `a if condition else b`
- `while` loop
- `for` loop (used most often)
- `continue` → skip to next round, `break` → stop the loop

```python
def grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 75:
        return "A"
    elif marks >= 50:
        return "B"
    else:
        return "Needs improvement"

for m in [95, 80, 62, 30]:
    print(m, "→", grade(m))

# One-line if-else
age = 17
status = "Can vote" if age >= 18 else "Cannot vote yet"
print(status)
```

```python
# while loop – saving ₹500 per week to buy a ₹2000 cricket bat
savings, week = 0, 0
while savings < 2000:
    week += 1
    savings += 500
print(f"Bat bought after {week} weeks")

# for loop with continue and break
for floor in range(1, 10):
    if floor == 4:
        continue        # skip floor 4 (lift doesn't stop here)
    if floor == 7:
        break           # top floor reached – stop
    print("Lift stops at floor", floor)
```

## Truthiness

- Booleans are `True` and `False`.
- `None` means "no value" (like `null` in other languages). Check it with `is None`, not `== None`.

**"Falsy" values** – Python treats all of these as `False` in an `if`:

`False`, `None`, `[]`, `{}`, `""`, `set()`, `0`, `0.0`

- `x or default` → use `default` if `x` is falsy
- `all([...])` → True if **every** item is truthy
- `any([...])` → True if **at least one** item is truthy

```python
x = None
print(x is None)    # Pythonic way

cart = []
if not cart:
    print("Your cart is empty!")

user_name = ""
display_name = user_name or "Guest"     # fallback value
print(display_name)

# all and any
documents_submitted = [True, True, False]   # Aadhaar, PAN, Photo
print("All documents submitted?", all(documents_submitted))
print("Any document submitted?", any(documents_submitted))
print(all([]), any([]))   # True, False – empty list special cases
```

## Sorting

- `sorted(x)` → returns a **new** sorted list (original unchanged)
- `x.sort()` → sorts the **same** list (in place)
- `reverse=True` → largest to smallest
- `key=...` → sort by some rule (a function)

```python
prices = [499, 99, 1299, 249]
print(sorted(prices))               # new list
print(prices)                       # unchanged

prices.sort(reverse=True)           # in place, high to low
print(prices)

# Sort by a rule: by length of city name
cities = ["Pune", "Bengaluru", "Agra", "Hyderabad", "Surat"]
print(sorted(cities, key=len))

# Sort a Counter-style list of (item, count) by count – most sold first
sales = {"Samosa": 40, "Vada Pav": 65, "Dosa": 52, "Idli": 30}
top_items = sorted(sales.items(), key=lambda pair: pair[1], reverse=True)
print(top_items)
```

## List comprehensions

A short, Pythonic way to **create a new list from an old one** – by filtering, changing, or both.

Pattern: `[expression for item in collection if condition]`



```python
marks = [35, 82, 47, 91, 66, 28]

passed = [m for m in marks if m >= 40]            # filter
print("Passed:", passed)

with_grace = [m + 5 for m in marks]               # transform
print("With 5 grace marks:", with_grace)

passed_with_grace = [m + 5 for m in marks if m >= 40]   # both
print(passed_with_grace)

# Dict comprehension – number : its square
squares = {n: n * n for n in range(1, 6)}
print(squares)

# Set comprehension
first_letters = {city[0] for city in ["Delhi", "Noida", "Mumbai", "Bhubaneswar"]}
print(first_letters)

# '_' when we don't need the value
zeros = [0 for _ in range(5)]
print(zeros)
```

```python
# Multiple 'for' in one comprehension – all shirt combinations
colours = ["red", "blue"]
sizes = ["S", "M", "L"]
combos = [(c, s) for c in colours for s in sizes]
print(combos)

# Later 'for' can use earlier ones – unique match pairs among 4 teams
teams = ["CSK", "MI", "KKR", "RCB"]
matches = [(teams[i], teams[j])
           for i in range(len(teams))
           for j in range(i + 1, len(teams))]
print(len(matches), "matches:", matches)
```

## Generators and iterators

A list keeps **all** its values in memory at once. If the list is huge (say 1 crore items), that wastes memory.

A **generator** gives values **one at a time, only when needed** (this is called *lazy* evaluation).

- Create with a function that uses `yield`
- Or with round brackets: `(x for x in ...)`
- A generator can be used **only once**.

```python
def token_numbers(start):
    #Like a bank token machine – gives the next number only when asked
    n = start
    while True:          # infinite generator!
        yield n
        n += 1

machine = token_numbers(101)
print(next(machine))   # 101
print(next(machine))   # 102
print(next(machine))   # 103

# Using it in a loop – always with a 'break' for infinite generators
for token in token_numbers(1):
    if token > 5:
        break
    print("Serving token", token)
```

```python
# Generator expression with ( )
even_squares = (n * n for n in range(10) if n % 2 == 0)
print(list(even_squares))   # [0, 4, 16, 36, 64]
print(list(even_squares))   # [] – already used up!

# Memory difference
import sys
big_list = [n for n in range(1_000_000)]
big_gen  = (n for n in range(1_000_000))
print("List size:", sys.getsizeof(big_list), "bytes")
print("Generator size:", sys.getsizeof(big_gen), "bytes")
```

## Randomness

The `random` module is used a lot in data science (simulations, sampling, splitting data).

| Function | What it does |
|---|---|
| `random.random()` | Random decimal between 0 and 1 |
| `random.seed(n)` | Fix the starting point → same "random" results every time (reproducible) |
| `random.randrange(a, b)` | Random integer from `range(a, b)` |
| `random.shuffle(lst)` | Shuffle a list in place |
| `random.choice(lst)` | Pick one item |
| `random.sample(lst, k)` | Pick `k` items **without** repeats |
| `random.choices(lst, k=k)` | Pick `k` items **with** repeats (Python 3) |


```python
import random

random.seed(42)
print([round(random.random(), 3) for _ in range(3)])

random.seed(42)                 # same seed again …
print([round(random.random(), 3) for _ in range(3)])   # … same numbers!

print("Dice roll:", random.randrange(1, 7))

players = ["Rohit", "Gill", "Kohli", "Pant", "Hardik"]
random.shuffle(players)
print("Batting order:", players)

print("Toss:", random.choice(["Heads", "Tails"]))

print("Lucky draw winners:", random.sample(range(1, 101), 3))      # no repeats
print("With repeats:", [random.choice(range(10)) for _ in range(5)])
```

## Regular expressions (regex)

Regex is a powerful way to **search and change text using patterns**. It is a big topic; here are just the basic functions from the `re` module:

| Function | Meaning |
|---|---|
| `re.match(p, s)` | Does `s` **start** with pattern `p`? |
| `re.search(p, s)` | Is `p` **anywhere** in `s`? |
| `re.split(p, s)` | Split `s` wherever `p` appears |
| `re.sub(p, r, s)` | Replace `p` with `r` in `s` |
| `re.findall(p, s)` | Get all matches as a list |

```python
import re

print(bool(re.match("K", "Kolkata")))       # True – starts with K
print(bool(re.match("o", "Kolkata")))       # False – doesn't start with o
print(bool(re.search("kat", "Kolkata")))    # True – 'kat' is inside

print(re.split("[,;]", "rice,dal;roti,sabzi"))          # split on , or ;
print(re.sub("[0-9]", "X", "My OTP is 482913"))          # hide digits

# Find all 10-digit mobile numbers in text
text = "Call 9876543210 or 9123456780 for booking."
print(re.findall(r"\b\d{10}\b", text))

# Simple check: is this a valid Indian PIN code (6 digits, not starting with 0)?
for pin in ["700091", "012345", "40001"]:
    print(pin, bool(re.fullmatch(r"[1-9]\d{5}", pin)))
```

## Classes

A **class** bundles **data** and the **functions that work on that data** together.

- Class names use **PascalCase** (e.g. `BankAccount`).
- `__init__` is the **constructor** – runs when we create a new object.
- `self` refers to the current object.
- `__repr__` decides how the object is shown when printed.


```python
class BankAccount:
    """A very simple bank account."""

    def __init__(self, holder, balance=0):
        """Constructor – called when a new account is created."""
        self.holder = holder
        self.balance = balance
        self.history = []

    def __repr__(self):
        return f"BankAccount({self.holder}, balance=₹{self.balance})"

    def deposit(self, amount):
        self.balance += amount
        self.history.append(("deposit", amount))

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
            return
        self.balance -= amount
        self.history.append(("withdraw", amount))

acc = BankAccount("Ravi", 5000)
acc.deposit(2000)
acc.withdraw(10000)     # not allowed
acc.withdraw(1500)
print(acc)
print(acc.history)
```

```python
class MySet:
    def __init__(self, values=None):
        self.dict = {}
        if values is not None:
            for v in values:
                self.add(v)
    def __repr__(self):
        return "MySet: " + str(list(self.dict.keys()))
    def add(self, value):
        self.dict[value] = True
    def contains(self, value):
        return value in self.dict
    def remove(self, value):
        del self.dict[value]

s = MySet(["red", "green", "red"])
s.add("blue")
s.remove("green")
print(s, s.contains("green"))
```

## Functional tools

### `functools.partial`
Creates a **new function by fixing some arguments** of an old function.

### `map`, `filter`, `reduce`
Alternatives to list comprehensions:
- `map(f, xs)` → apply `f` to every item
- `filter(f, xs)` → keep items where `f` is True
- `reduce(f, xs)` → combine all items into **one** result, step by step

```python
from functools import partial, reduce

def price_after_discount(price, discount_percent):
    return price * (100 - discount_percent) / 100

# partial – make fixed-discount functions
diwali_sale = partial(price_after_discount, discount_percent=30)
clearance   = partial(price_after_discount, discount_percent=50)
print(diwali_sale(2000)) 
print(clearance(2000))    
```
