## Arithmetic
- `/` → normal division (gives decimal)
- `//` → integer (floor) division
- `%` → remainder
- `**` → power

```python
print(5 / 2)    # normal division
print(5 // 2)   # integer division
print(5 % 2)    # remainder
print(2 ** 10)  # Power

# example: 10 samosas shared among 3 friends
samosas, friends = 10, 3
print("Each friend gets:", samosas // friends, "samosas")
print("Left over:", samosas % friends)
```

## Functions

A **function** takes zero or more inputs and gives back an output. We define it with `def`.

Key points:
- A **docstring** (text in triple quotes) explains what the function does.
- Functions are **first-class** – you can store them in variables and pass them to other functions.
- **Lambda** = a short, one-line function without a name.
- Parameters can have **default values**.
- You can pass arguments **by name**.

```python
def add_gst(price):
    """Returns the price after adding 18% GST."""
    return price * 1.18

print(add_gst(1000))
```

```python
# Functions are first-class: we can pass one function into another
def apply_to_100(f):
    """Calls the function f with 100 as input."""
    return f(100)

my_func = add_gst         
print(apply_to_100(my_func))   

# Lambda – a small nameless function
print(apply_to_100(lambda x: x - 10))

# Note: assigning a lambda to a name works, but 'def' is preferred
def half(x): return x / 2
```

```python
# Default arguments
def greet(name="Guest"):
    print(f"Namaste, {name}!")

greet("Debasis Das")
greet()          

# Arguments by name
def emi_left(total=0, paid=0):
    return total - paid

print(emi_left(50000, 20000))
print(emi_left(paid=5000))  
```

## Exceptions

When something goes wrong, Python **raises an exception**. If we don't handle it, the program crashes.

We handle it using `try` and `except`. In Python, using exceptions to keep code clean is perfectly normal.

```python
def per_head_cost(total_bill, people):
    try:
        return total_bill / people
    except ZeroDivisionError:
        print("Cannot divide the bill among 0 people!")
        return None

print(per_head_cost(1200, 4))
print(per_head_cost(1200, 0))
```

## Lists

A **list** is an ordered collection – the most important data structure in Python. (Like an array in other languages, but more powerful.)

| Operation | Example |
|---|---|
| Length / Sum | `len(x)`, `sum(x)` |
| Index (starts from 0) | `x[0]`, last item `x[-1]` |
| Slicing | `x[1:4]`, `x[:3]`, `x[-2:]` |
| Check membership | `5 in x` (slow for big lists) |
| Join lists | `x.extend([...])` or `x + [...]` |
| Add one item | `x.append(item)` |
| Unpacking | `a, b = [1, 2]` |

```python
runs = [45, 12, 78, 101, 33, 0, 56]   # runs scored in 7 matches

print("Matches played:", len(runs))
print("Total runs:", sum(runs))
print("First match:", runs[0])
print("Last match:", runs[-1])
print("Second last:", runs[-2])

print("First three:", runs[:3])
print("From 4th match:", runs[3:])
print("Matches 2 to 4:", runs[1:4])
print("Last two:", runs[-2:])

print("Any century?", 101 in runs)
print("Any 200?", 200 in runs)
```

```python
# Mixed data types are allowed in one list
student = ["Rahul", 21, 8.7, True]

# Joining and adding
fruits = ["mango", "banana"]
fruits.extend(["guava", "papaya"])      # changes fruits
print(fruits)

more_fruits = fruits + ["litchi"]       # new list, fruits unchanged
print(more_fruits)

fruits.append("jackfruit")              # add one item
print(fruits)

# Unpacking
first, second = ["Delhi", "Pune"]
print(first, second)

_, capital = ["India", "New Delhi"]     # '_' means we don't need that value
print(capital)
```

## Tuples

A **tuple** is like a list, but it **cannot be changed** (it is *immutable*). We write it with round brackets `( )`.

Tuples are very handy for:
- Returning **multiple values** from a function.
- **Swapping** variables in one line.

```python
train = ("12951", "Bhubaneswar Rajdhani")   # tuple
try:
    train[1] = "Duronto"
except TypeError:
    print("Cannot modify a tuple!")

def min_and_max(temps):
    return min(temps), max(temps)      # returns a tuple

low, high = min_and_max([28, 31, 35, 33, 29])
print(f"Lowest: {low}°C, Highest: {high}°C")

# Swapping – the Pythonic way
a, b = "tea", "coffee"
a, b = b, a
print(a, b)   # coffee tea
```

## Dictionaries

A **dictionary** stores **key → value** pairs. You can find a value very fast using its key.

- Look up: `d[key]` (gives `KeyError` if key is missing)
- Safe look up: `d.get(key, default)`
- Check key: `key in d`
- Add / update: `d[key] = value`
- `.keys()`, `.values()`, `.items()` give all keys, values, or pairs.
- Keys must be **immutable** (string, number, tuple – **not** list).

```python
pin_codes = {"Kolkata": 700001, "Mumbai": 400001, "Chennai": 600001}

print(pin_codes["Mumbai"])

try:
    print(pin_codes["Jaipur"])
except KeyError:
    print("No PIN code saved for Jaipur!")

print("Chennai" in pin_codes)              # True
print(pin_codes.get("Jaipur", "Unknown"))  # Unknown – no error

pin_codes["Jaipur"] = 302001               # add new
pin_codes["Kolkata"] = 700091              # update existing
print(len(pin_codes), pin_codes)
```

```python
# Dictionaries are great for structured data – like one record
movie = {
    "title": "Dhurandar",
    "director": "Aditya Dhar",
    "year": 2023,
    "languages": ["Hindi", "English", "Tamil", "Telugu"],
}

print(movie.keys())
print(movie.values())
for key, value in movie.items():
    print(f"{key:>10} : {value}")

print("director" in movie)  # True
```

## `defaultdict`

Suppose we want to **count words**. With a normal dict we must first check *"is this key already there?"* every time. That is boring.

A **`defaultdict`** solves this. If a key is missing, it automatically creates it with a default value:
- `defaultdict(int)` → default `0`
- `defaultdict(list)` → default `[]`
- `defaultdict(dict)` → default `{}`

```python
sentence = "chai pe charcha chai pe samosa chai"
words = sentence.split()

# Way 1: normal dict – need an if/else check
counts = {}
for w in words:
    if w in counts:
        counts[w] += 1
    else:
        counts[w] = 1
print("Normal dict:", counts)

# Way 2: dict.get
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1
print("Using get:  ", counts)

# Way 3: defaultdict – cleanest
from collections import defaultdict
counts = defaultdict(int)
for w in words:
    counts[w] += 1
print("defaultdict:", dict(counts))
```

```python
# defaultdict(list) – group students by their city
students = [("Amit", "Delhi"), ("Ricky", "Bhubaneswar"), ("Ravi", "Delhi"), ("Anu", "Pune"), ("Kiran", "Goa")]

by_city = defaultdict(list)
for name, city in students:
    by_city[city].append(name)
print(dict(by_city))

# defaultdict(dict) – nested data without checks
profile = defaultdict(dict)
profile["Amit"]["city"] = "Delhi"
print(dict(profile))
```

### `Counter`

**`Counter`** turns a list of items into a dictionary of **item → count**. It is the easiest way to make a frequency table (histogram).

`most_common(n)` gives the top `n` items.

```python
from collections import Counter

# Class poll: "Which is your favourite sweet?"
votes = ["Rasgulla", "Jalebi", "Ladoo", "Rasgulla", "Gulab Jamun",
         "Jalebi", "Rasgulla", "Ladoo", "Rasgulla"]
vote_count = Counter(votes)
print(vote_count)

for sweet, count in vote_count.most_common(2):
    print(sweet, count)

# Word counting in one line
print(Counter(sentence.split()).most_common(1))
```
