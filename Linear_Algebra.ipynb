# Linear Algebra

## Vector
  
A **vector** is an ordered list of numbers. Each number describes one feature of something.

 - The number of values in a vector is called its **dimension**.
 - **Scalar** is a only number but **vector** is both define number and direction.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Example list of height, weight, and age of 5 individuals

height = [160, 150, 170, 140, 120]
weight = [60, 50, 70, 40, 30]
age = [25, 30, 22, 28, 23]

# Each person becomes one vector: [height, weight, age]
people = [[h, w, a] for h, w, a in zip(height, weight, age)]

for i, person in enumerate(people, start=1):
    print(f"Person {i}: {person}")
```

```python
# Vector dimensions

people_dim = len(people[0])  
print(f"Each person vector has {people_dim} dimensions (height, weight, age).")
```

```python
# Vector Addition and Subtraction

a = [5, 10, 12, 17, 21, 25]
b = [2, 4, 6, 8, 10, 12]
c = [7, 14, 18, 25, 31, 37]

vector_sum = [x + y + z for x, y, z in zip(a, b, c)]
print("Vector sum:", vector_sum)

vector_diff = [x - y for x, y in zip(a, b)]
print("Vector difference:", vector_diff)
```

```python
# Vector multiplication by a scalar

vector_mult = [2 * x for x in a]
print("Vector multiplied by scalar 2:", vector_mult)
```

```python
# vector mean

vector_mean = [sum(a) / len(a), sum(b) / len(b), sum(c) / len(c)]
print("Vector mean:", vector_mean)

vec_mean = np.mean([a, b, c], axis=1)
print("Vector mean using numpy:", vec_mean)
```

```python
# vector dot product

dot_product_ab = sum(x * y for x, y in zip(a, b))
print("Dot product of vectors a and b:", dot_product_ab)

dot_product_ac = sum(x * y for x, y in zip(a, c))
print("Dot product of vectors a and c:", dot_product_ac)

dot_product_bc = sum(x * y for x, y in zip(b, c))
print("Dot product of vectors b and c:", dot_product_bc)
```

```python
# Vector sum of squares

sum_of_squares_a = sum(x ** 2 for x in a)
print("Sum of squares of vector a:", sum_of_squares_a)

sum_of_squares_b = sum(x ** 2 for x in b)
print("Sum of squares of vector b:", sum_of_squares_b)

sum_of_squares_c = sum(x ** 2 for x in c)
print("Sum of squares of vector c:", sum_of_squares_c)
```

```python
# vector magnitude

a_mag = np.sqrt(sum_of_squares_a)
print("Magnitude of vector a:", a_mag)

b_mag = np.sqrt(sum_of_squares_b)
print("Magnitude of vector b:", b_mag)  

c_mag = np.sqrt(sum_of_squares_c)
print("Magnitude of vector c:", c_mag)
```

## Matrix

A **matrix** is a table of numbers arranged in rows (going across) and columns (going down).

- Each **row** is usually one item (for example, one person).
- Each **column** is usually one feature (for example, height).
- A matrix is simply many vectors put together.
- Shape of matrix is `rows x column` (for example 5 x 3)
- If **A** is matrix, then each elements of matrix represent as `A[i][j]`,where i is row and j is column

```python
# Genrate a random matrix of size 3x3
random_matrix = np.random.rand(3, 3)
print("Random 3x3 matrix:")
print(random_matrix)
```

```python
# 5 x 5 identity matrix

identity_matrix = np.eye(5)
print("5x5 Identity matrix:")   
print(identity_matrix)
```

```python
# Determinant of a matrix
determinant = np.linalg.det(random_matrix)
print("Determinant of the random 3x3 matrix:", determinant) 
```

```python
# Inverse of a matrix
inverse_matrix = np.linalg.inv(random_matrix)
print("Inverse of the random 3x3 matrix:")  
print(inverse_matrix)
```

```python
# Transpose of a matrix

transpose_matrix = np.transpose(random_matrix)
print("Transpose of the random 3x3 matrix:")    
print(transpose_matrix)
```
