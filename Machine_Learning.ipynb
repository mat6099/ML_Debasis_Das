# Machine Learning

## What is Machine Learning ?

- **Machine learning** means building models that are learned from data, so that they can predict answer for new data i.e machine learn patterns from the data and replicate same in future.
- It is also called **predictive modelling** or **data mining**. We use old data to build a model, then use it on new data.

### Example:

- Predicting email message is spam or not ?
- Predicting fraud credict card transanction
- Predicting House price in a area of cities
- Predicting cricket team win or lose
- Predicting which related product customer buy with main product 

## Types Of Machine Learning:

1. **Supervised Machine Learning**: Data Comes with the correct answer i.e historical data act as supervise. Example, predicting email is spam or not.
2. **Unsupervised Machine Learning**: The model finds patterns by itself i.e no historical data. Example, Grouping customers by shopping habits.
3. **Semisupervised Machine Learning**: Some of data has labels i.e it is mixture of supervised and unsupervised machine learning. Example, A few photos tagged, most untagged.
4. **Reinforcement Learning**: Keep learning from as new data keep arriving i.e it learn from its experience. Example, A news aap that learns customer taste every day.
